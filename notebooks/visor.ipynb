# Visor de datos COFEPRIS BRSDM

Esta notebook ofrece una vista exploratoria rápida de los datos cargados por el proyecto `cofepris-BRSDM-scraper`.

Objetivo:
- conectarse a PostgreSQL,
- cargar la tabla de medicamentos en un DataFrame de pandas,
- inspeccionar estadísticas generales y filas de ejemplo,
- sugerir análisis visuales útiles.

```python
#@title Librerías
import os

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import userdata
from sqlalchemy import create_engine, text
```

## Variables de entorno (SECRETS)

Si estás en colab, en el menú en secrets deben agregarse las variables de entorno **DATABASE_URL**

Por default esta variables de entorno contiene la url y el nombre de la tabla del contenedor creado en dev al ejecutar el proyecto, si utilizas otro servidor de base de datos agrégala.



```python
#@title Creación del Dataframe

DATABASE_URL = userdata.get('DATABASE_URL')
TABLE_NAME = "medicamentos"

engine = create_engine(DATABASE_URL)
query = text(f"SELECT * FROM {TABLE_NAME}")

df = pd.read_sql(query, engine)

print(f"Loaded {len(df):,} rows from table '{TABLE_NAME}'.")
df.shape
```

```python
#@title Descripción estadística de los datos
df.describe(include="all").transpose()
```

## Preprocesamiento de Datos

Antes de graficar, realizaremos un preprocesamiento de las columnas relevantes:
- Convertir `fecha_emision` a tipo datetime.
- Llenar valores nulos en columnas categóricas con 'Desconocido' para asegurar que aparezcan en las gráficas, o simplemente eliminar nulos antes del conteo.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convertir 'fecha_emision' a datetime, forzando errores a NaT
df['fecha_emision'] = pd.to_datetime(df['fecha_emision'], errors='coerce')

# Rellenar NaNs en columnas categóricas importantes con 'Desconocido' para la visualización
# O simplemente eliminar NaNs para el conteo de frecuencias
for col in ['sustancia_quimica', 'estado', 'grupo_farmacologico', 'titular', 'presentacion', 'forma_farmaceutica']:
    if col in df.columns:
        # For frequency counts, it's often better to drop NaNs or count them explicitly
        # For now, we'll just ensure they are of the correct type if not already.
        # If we fill, it would be df[col] = df[col].fillna('Desconocido')
        pass

print("Preprocesamiento de fechas y manejo básico de nulos completado.")
```

---

## 1. Principales sustancias activas (`sustancia_quimica`)
Gráfica de barras horizontal con los 20 valores más frecuentes.

```python
plt.figure(figsize=(12, 8))
# Contar las frecuencias y tomar los top 20, excluyendo NaNs
top_sustancias = df['sustancia_quimica'].value_counts().nlargest(20)
sns.barplot(x=top_sustancias.values, y=top_sustancias.index, palette='viridis')
plt.title('Top 20 Sustancias Químicas Más Frecuentes')
plt.xlabel('Número de Registros')
plt.ylabel('Sustancia Química')
plt.tight_layout()
plt.show()
```

---

## 2. Distribución por estatus (`estado`)
Gráfica de barras para comparar `VIGENTE`, `CANCELADO`, etc.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='estado', data=df, palette='rocket', order=df['estado'].value_counts().index)
plt.title('Distribución de Registros por Estado')
plt.xlabel('Estado')
plt.ylabel('Número de Registros')
plt.tight_layout()
plt.show()
```

---

## 3. Desglose por grupo terapéutico (`grupo_farmacologico`)
Gráfica de barras de los grupos farmacológicos más frecuentes.

```python
plt.figure(figsize=(12, 8))
# Contar las frecuencias y tomar los top 20, excluyendo NaNs
top_grupos = df['grupo_farmacologico'].value_counts().nlargest(20)
sns.barplot(x=top_grupos.values, y=top_grupos.index, palette='mako')
plt.title('Top 20 Grupos Farmacológicos Más Frecuentes')
plt.xlabel('Número de Registros')
plt.ylabel('Grupo Farmacológico')
plt.tight_layout()
plt.show()
```

---

## 4. Tendencia de emisión en el tiempo (`fecha_emision`)
Línea mensual después de parsear fechas.

```python
plt.figure(figsize=(15, 7))
# Agrupar por mes y contar registros, eliminando NaNs de fecha_emision
df_time_series = df.set_index('fecha_emision').resample('M').size().reset_index(name='count')

sns.lineplot(x='fecha_emision', y='count', data=df_time_series)
plt.title('Tendencia Mensual de Emisión de Registros')
plt.xlabel('Fecha de Emisión')
plt.ylabel('Número de Registros')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

---

## 5. Titulares más frecuentes (`titular`)
Top 15 entidades por número de registros.

```python
plt.figure(figsize=(12, 8))
# Contar las frecuencias y tomar los top 15, excluyendo NaNs
top_titulares = df['titular'].value_counts().nlargest(15)
sns.barplot(x=top_titulares.values, y=top_titulares.index, palette='cubehelix')
plt.title('Top 15 Titulares Más Frecuentes')
plt.xlabel('Número de Registros')
plt.ylabel('Titular')
plt.tight_layout()
plt.show()
```
